```python
import pandas as pd
import time
from google.colab import files


# Start timing
start_time = time.time()

df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/clean_data.csv")
# Optional: Preview the data
df.head()
def explain_churn(row):
    reasons = []

    # Contract type
    if "Contract_Encoded" in row and row["Contract_Encoded"] == 0:
        reasons.append("The customer is on a month-to-month contract, which often leads to higher churn risk.")

    # Tenure
    if "tenure_<12" in row and row["tenure_<12"] == 1:
        reasons.append("The customer has been with the company for less than a year, suggesting lower loyalty.")
    elif "tenure_>60" in row and row["tenure_>60"] == 1:
        reasons.append("The customer has been with the company for a long time, indicating strong retention potential.")

    # Internet service type
    if "InternetService_Fiber optic" in row and row["InternetService_Fiber optic"] == 1:
        reasons.append("Fiber optic users sometimes face higher bills, which may cause dissatisfaction.")
    elif "InternetService_No" in row and row["InternetService_No"] == 1:
        reasons.append("The customer does not have internet service, which might indicate limited engagement.")

    # Technical support and security
    if "TechSupport" in row and row["TechSupport"] == "No":
        reasons.append("The customer lacks tech support, which may increase frustration with technical issues.")
    if "OnlineSecurity" in row and row["OnlineSecurity"] == "No":
        reasons.append("The customer lacks online security, which could reduce perceived value.")

    # Billing and payment
    if "PaperlessBilling" in row and row["PaperlessBilling"] == "Yes":
        reasons.append("The customer uses paperless billing, but that doesn’t directly affect churn.")
    if "AutomaticPayment" in row and row["AutomaticPayment"] == 0:
        reasons.append("The customer does not use automatic payments, which may indicate lower engagement.")

    # Charges
    if "MonthlyCharges" in row and row["MonthlyCharges"] > 80:
        reasons.append("The customer has high monthly charges, which can contribute to dissatisfaction.")
    elif "MonthlyCharges" in row and row["MonthlyCharges"] < 40:
        reasons.append("The customer has low monthly charges, which could indicate a basic plan and lower switching cost.")

    # Final sentiment
    if row["Churn"] == 1:
        summary = "This customer **churned** likely due to: " + " ".join(reasons)
    else:
        summary = "This customer **did not churn**, likely due to good retention factors such as long tenure or favorable contract terms."

    return summary
# Apply the explanation generator to every row
df["Churn_Explanation"] = df.apply(explain_churn, axis=1)
# End timing
end_time = time.time()

# Calculate elapsed time
elapsed = end_time - start_time
print(f"⏱️ Benchmark: Process completed in {elapsed:.2f} seconds")
# View sample outputs
df[["gender", "Contract_Encoded", "MonthlyCharges", "Churn", "Churn_Explanation"]].head(10)
df.to_csv("customer_churn_explanations.csv", index=False)
```
